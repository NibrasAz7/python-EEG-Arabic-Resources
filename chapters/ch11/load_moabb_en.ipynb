## Overview

MOABB (Mother of All BCI Benchmarks) provides standardized access to public BCI datasets. This notebook loads the BNCI2014-001 motor imagery dataset for subject 1, extracts epochs with the MotorImagery paradigm, and visualizes the number of trials per class.

## What this notebook does

- Loads BNCI2014-001 subject 1 via MOABB (data downloads automatically on first run)
- Extracts epochs using the MotorImagery paradigm
- Prints dataset structure: sessions, runs, channels, sampling rate
- Counts trials per class and plots a bar chart

## What you should expect to see

- 2 sessions with 6 runs each
- 22 EEG channels at 250 Hz
- 576 total epochs across 4 classes (144 each)
- A bar chart showing balanced trial counts

## Key parameters

| Parameter | Value | Meaning |

| dataset | BNCI2014_001 | MOABB motor imagery dataset |
| subjects | [1] | Subject 1 only |

## 2. Load MOABB dataset

MOABB downloads data automatically on first use (~44 MB for subject 1). Subsequent runs use cached data.



```python
from moabb.datasets import BNCI2014_001
ds = BNCI2014_001()
sessions = ds.get_data(subjects=[1])
subject_key = list(sessions.keys())[0]
session_dict = sessions[subject_key]
n_sessions = len(session_dict)
n_runs = len(next(iter(session_dict.values())))
first_run = next(iter(next(iter(session_dict.values())).values()))
n_channels_raw = len(first_run.ch_names)
sfreq = first_run.info['sfreq']
print(f'Subject 1: {n_sessions} sessions, {n_runs} runs/session')
print(f'Raw channels: {n_channels_raw}, Sampling rate: {sfreq} Hz')
print(f'Channel names: {first_run.ch_names}')

```

```python
from moabb.paradigms import MotorImagery
paradigm = MotorImagery(n_classes=2)
X, labels, meta = paradigm.get_data(dataset=ds, subjects=[1])
print(f'X shape: {X.shape}  (n_trials, n_channels, n_samples)')
print(f'Labels shape: {labels.shape}')
print(f'Meta shape: {meta.shape}')

```

## 3. Explore the data

We print the epoch shapes, labels, and trial counts per class.



```python
import numpy as np
unique_labels, counts = np.unique(labels, return_counts=True)
print(f'Epoch channels: {X.shape[1]}')
print(f'Epoch samples: {X.shape[2]}')
print(f'Epoch duration: {X.shape[2] / sfreq:.2f} s')
print(f'Unique labels: {list(unique_labels)}')
print(f'Trials per class: {dict(zip(unique_labels, counts))}')
print(f'Total trials: {X.shape[0]}')

```

## 4. Apply the analysis

We count trials per class to prepare for the bar chart.



```python
import numpy as np
unique_labels, counts = np.unique(labels, return_counts=True)
trial_counts = dict(zip(unique_labels, counts))
print(f'Trial counts: {trial_counts}')

```

## 5. Interactive plot

**What to look for:**

- All four classes have equal trial counts (balanced dataset)
- The bar chart shows 144 trials per class
- Hover over bars to see exact values




```python
import plotly.graph_objects as go
fig = go.Figure(data=[go.Bar(x=list(trial_counts.keys()), y=list(trial_counts.values()),
                             marker_color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'],
                             text=list(trial_counts.values()), textposition='auto')])
fig.update_layout(title='Trials per class - BNCI2014-001 (subject 1)',
                  xaxis_title='Class label', yaxis_title='Number of trials',
                  height=500)
fig.show()

```
