## 3. التصويرُ الثلاثيُّ الأبعاد

### علامَ تلاحظ؟
- توزيعَ السعاتِ عبرَ القنواتِ الأربع
- الفروقَ في المدى بينَ القنوات
- التقليصَ لا يُؤثّرُ على الشكلِ العامّ

</div>

```python
import sys
sys.path.insert(0, '.')
import numpy as np
import plotly.graph_objects as go
from utils.eeg_loader import load_local_eeg

timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local', subject=7, experiment=1, session=2
)

fs = 200
step = 10  # subsample for clarity
time_sec = np.arange(0, eeg_data.shape[0], step) / fs
data_sub = eeg_data[::step, :]

fig = go.Figure()
for i, ch in enumerate(ch_names):
    fig.add_trace(go.Scatter3d(
        x=time_sec, y=np.full_like(time_sec, i), z=data_sub[:, i],
        mode='markers', marker=dict(size=1, color=data_sub[:, i],
                                    colorscale='viridis'),
        name=ch
    ))

fig.update_layout(
    scene=dict(
        xaxis_title='Time (s)', yaxis_title='Channel',
        zaxis_title='Amplitude (uV)',
        yaxis=dict(tickvals=list(range(len(ch_names))), ticktext=ch_names)
    ),
    title='3D EEG Visualization - Subject 7'
)
fig.show()
```
