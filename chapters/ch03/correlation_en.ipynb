## 3. Compute and plot correlation matrix

### What to look for
- Values on the main diagonal (1.000) since each channel perfectly correlates with itself
- High correlations between channels (indicating shared components)
- Lowest correlation between Cz and F8

```python
import sys
sys.path.insert(0, '.')
import numpy as np
import plotly.graph_objects as go
from utils.eeg_loader import load_local_eeg

timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local', subject=7, experiment=1, session=2
)

corr_matrix = np.corrcoef(eeg_data.T)

fig = go.Figure(data=go.Heatmap(
    z=corr_matrix, x=ch_names, y=ch_names,
    colorscale='RdBu_r', zmin=-1, zmax=1,
    text=[[f'{corr_matrix[i, j]:.2f}' for j in range(len(ch_names))]
          for i in range(len(ch_names))],
    texttemplate='%{text}', colorbar=dict(title='Correlation')
))
fig.update_layout(title='EEG Channel Correlation Matrix')
fig.show()

print('Correlation matrix:')
print(f"{'':>6}", end='')
for ch in ch_names:
    print(f'{ch:>8}', end='')
print()
for i, ch in enumerate(ch_names):
    print(f'{ch:>6}', end='')
    for j in range(len(ch_names)):
        print(f'{corr_matrix[i, j]:>8.3f}', end='')
    print()
```

## Summary

- High correlations in raw data are common due to drift and shared noise
- Correlations typically decrease after filtering and artifact removal
- Highest correlation is between P4 and T7 (0.91)
