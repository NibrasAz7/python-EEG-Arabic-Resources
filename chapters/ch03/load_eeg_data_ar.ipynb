## 3. تحميلُ البياناتِ وعرضُ المعلومات

نَحمّلُ المُشاركَ 7، التجربةَ 1، الجلسةَ 2. هذه تجربةٌ سمعيّةٌ تَستمرُّ نحوَ 158 ثانية.

```python
import sys
sys.path.insert(0, '.')
from utils.eeg_loader import load_local_eeg

# Load subject 7, experiment 1, session 2
timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local',
    subject=7,
    experiment=1,
    session=2
)

print(f'Signals shape: {eeg_data.shape}')
print(f'Channels: {ch_names}')
print(f'Duration: {eeg_data.shape[0] / 200:.1f} seconds')
print(f'Sampling rate: 200 Hz')

for i, ch in enumerate(ch_names):
    signal = eeg_data[:, i]
    print(f'  {ch}: mean={signal.mean():.2f} uV, '
          f'std={signal.std():.2f} uV, '
          f'range=[{signal.min():.1f}, {signal.max():.1f}] uV')
```
