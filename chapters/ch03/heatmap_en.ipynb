## 3. Create heatmap

### What to look for
- Time regions with high amplitudes (more saturated colors)
- Drift in channel P4 as a continuous blue region
- Synchronization between channels at specific moments

```python
import sys
sys.path.insert(0, '.')
import matplotlib.pyplot as plt
from utils.eeg_loader import load_local_eeg

timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local', subject=7, experiment=1, session=2
)

fig, ax = plt.subplots(figsize=(12, 4))
im = ax.imshow(
    eeg_data.T, aspect='auto', cmap='RdBu_r',
    vmin=-100, vmax=100,
    extent=[0, eeg_data.shape[0]/200, 4, 0]
)
ax.set_yticks([0.5, 1.5, 2.5, 3.5])
ax.set_yticklabels(ch_names)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Channel')
ax.set_title('EEG Heatmap - Subject 7')
plt.colorbar(im, ax=ax, label='Amplitude (uV)')
plt.tight_layout()
plt.show()
```

## Summary

- Heatmap summarizes amplitude across channels and time
- RdBu_r colormap: red for positive, blue for negative
- Reveals temporal patterns shared across channels
