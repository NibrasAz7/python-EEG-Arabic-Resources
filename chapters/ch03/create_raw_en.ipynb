## 3. Create Raw object

Note: MNE expects data in shape (n_channels, n_samples), so we use eeg_data.T to transpose.

```python
import sys
sys.path.insert(0, '.')
import mne
from utils.eeg_loader import load_local_eeg

timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local', subject=7, experiment=1, session=2
)

info = mne.create_info(ch_names=ch_names, sfreq=200, ch_types='eeg')
raw = mne.io.RawArray(eeg_data.T, info)

print(raw)
print(f'Channels: {raw.ch_names}')
print(f'Duration: {raw.times[-1]:.1f} s')
```

## Summary

- Raw object stores the signal with metadata (channel names, sampling rate)
- MNE expects shape (channels, samples) while load_local_eeg returns (samples, channels)
- We use .T to transpose the array
