## 3. Plot topographic map

### What to look for
- Distribution of spectral power across the 4 electrode positions
- The map is approximate because 4 channels are insufficient for accurate spatial interpolation
- Clinical studies use 32 or 64 channels for higher precision

```python
import sys
sys.path.insert(0, '.')
import matplotlib.pyplot as plt
import mne
from utils.eeg_loader import load_local_eeg

timestamps, eeg_data, ch_names = load_local_eeg(
    data_dir='data/local', subject=7, experiment=1, session=2
)

info = mne.create_info(ch_names=ch_names, sfreq=200, ch_types='eeg')
raw = mne.io.RawArray(eeg_data.T, info)

montage = mne.channels.make_standard_montage('standard_1020')
raw.set_montage(montage, on_missing='ignore')

fig = raw.compute_psd().plot_topomap()
plt.tight_layout()
plt.show()
```
